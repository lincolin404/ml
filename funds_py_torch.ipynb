```python
import torch

tensor_a = torch.arange(3, 13)
print(tensor_a.shape)

x = tensor_a.reshape(5, 2)
#print(x)

tensor_b = torch.arange(1,11)
print(tensor_b, tensor_a)
stacked = torch.stack([tensor_a, tensor_b], dim=1)
print(stacked)
x = stacked.reshape(1,5,4)
print("..................................")
print(x)
print(x.shape)
y = torch.squeeze(x)
print("..................................")
print(y)
print(y.shape)
print("..................................")
z = y.unsqueeze(dim=1)
print(z)
print(z.shape)
print("..................................")
w = torch.randn(2,3,4)
print(w)
print(w.shape)
torch.permute(w, (2,0,1))
print(w)
print(w.shape)


```
